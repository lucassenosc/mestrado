```python
import pandas as pd
import os
import numpy as np
from sklearn.model_selection import train_test_split

# ==========================================
# 1. CONFIGURAÇÕES
# ==========================================
# Substitua pelo caminho e nome do seu ficheiro original
CAMINHO_DATASET_ORIGINAL = "data3/cicids2017_cleaned.csv" 
PASTA_SAIDA = "data3"

# ==========================================
# 2. CARREGAMENTO DOS DADOS
# ==========================================
print(f"A ler o dataset original: {CAMINHO_DATASET_ORIGINAL}...")
df = pd.read_csv(CAMINHO_DATASET_ORIGINAL)

print(f"Dataset carregado com {df.shape[0]} linhas e {df.shape[1]} colunas.")

# Remover espaços em branco nos nomes das colunas
df.columns = df.columns.str.strip()

# ==========================================
# 3. TRANSFORMAÇÃO DAS COLUNAS ALVO
# ==========================================
print("A gerar a coluna binária 'label' e a formatar 'attack_cat'...")

# Converter tudo para minúsculas e retirar espaços para evitar falhas por formatação
df['Attack Type'] = df['Attack Type'].astype(str).str.strip().str.lower()

# A sua string original é 'Normal Traffic'. Em minúsculas fica 'normal traffic'
TERMO_NORMAL = 'normal traffic'

# Criar a coluna 'label' (Binária: 0 para Normal, 1 para Ataque)
df['label'] = df['Attack Type'].apply(lambda x: 0 if x == TERMO_NORMAL else 1)

# Renomear 'Attack Type' para 'attack_cat' (a coluna que o nosso código principal procura)
df.rename(columns={'Attack Type': 'attack_cat'}, inplace=True)

# Imprimir um pequeno resumo para confirmarmos que as duas classes existem!
contagem_binaria = df['label'].value_counts()
print("\n--- Distribuição da Classe Binária ---")
print(f"Classe 0 (Normal): {contagem_binaria.get(0, 0)} pacotes")
print(f"Classe 1 (Ataque): {contagem_binaria.get(1, 0)} pacotes")

# Limpar valores NaN e Infinitos
df.fillna(0, inplace=True)
df.replace([np.inf, -np.inf], 0, inplace=True)

# ==========================================
# 4. DIVISÃO TREINO E TESTE (80% / 20%)
# ==========================================
print("\nA dividir os dados em Treino e Teste (Estratificado)...")
df_train, df_test = train_test_split(df, test_size=0.20, random_state=42, stratify=df['attack_cat'])

# ==========================================
# 5. GUARDAR OS FICHEIROS
# ==========================================
os.makedirs(PASTA_SAIDA, exist_ok=True)

caminho_treino = os.path.join(PASTA_SAIDA, "CICIDS_training-set.csv")
caminho_teste = os.path.join(PASTA_SAIDA, "CICIDS_testing-set.csv")

print(f"A guardar conjunto de Treino em {caminho_treino}...")
df_train.to_csv(caminho_treino, index=False)

print(f"A guardar conjunto de Teste em {caminho_teste}...")
df_test.to_csv(caminho_teste, index=False)

print("\n🚀 Transformação concluída! Pode correr a Standard WiSARD agora.")
```
